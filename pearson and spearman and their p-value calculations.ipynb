```python
# @title
import os
import json
import pandas as pd
from google.colab import drive

# =====================================================
# 1) اتصال به Google Drive
# =====================================================
drive.mount('/content/drive')

# =====================================================
# 2) پوشه‌های مدل‌ها
# =====================================================
models =["TimesFM", "Timer", "Moirai", "Lag_Llama", "Chronos","AutoETS","SNaive","AutoARIMA"]

base_root = "/content/drive/MyDrive/results/paper/json"

# =====================================================
# 3) لیست نتایج
# =====================================================
rows = []

# =====================================================
# 4) پردازش فایل‌های JSON
# =====================================================
for model in models:
    model_path = os.path.join(base_root, model)

    if not os.path.isdir(model_path):
        print(f"❌ پوشه پیدا نشد: {model}")
        continue

    json_files = [f for f in os.listdir(model_path) if f.endswith(".json")]
    print(f"📁 {model} | {len(json_files)} فایل")

    for file_name in json_files:
        file_path = os.path.join(model_path, file_name)

        try:
            with open(file_path, "r") as f:
                data = json.load(f)

            if len(data) == 0:
                continue

            run = data[-1]   # آخرین اجرا
            metrics = run.get("metrics", {})

            rows.append({
                "model": model,
                "file_name": file_name,
                "mae_scaled": metrics.get("mae_scaled"),
                "rmse_scaled": metrics.get("rmse_scaled"),
                "mse_scaled": metrics.get("mse_scaled"),
                "crps": metrics.get("crps")
            })

        except Exception as e:
            print(f"⚠️ خطا | {model}/{file_name}: {e}")

# =====================================================
# 5) ساخت DataFrame
# =====================================================
df_metrics = pd.DataFrame(rows)

# =====================================================
# 6) اضافه کردن ADF (بر اساس فایل)
# =====================================================

df_metrics["ADF"] = df_metrics["file_name"].map(adf_values)
df_metrics["FCA"] = df_metrics["file_name"].map(f_values)


# =====================================================
# 7) ساخت dataset_id بر اساس ترتیب الفبا
# =====================================================
dataset_order = sorted(df_metrics["file_name"].unique())

dataset_id_map = {
    name: idx + 1
    for idx, name in enumerate(dataset_order)
}

df_metrics["dataset_id"] = df_metrics["file_name"].map(dataset_id_map)

# =====================================================
# 8) مرتب‌سازی نهایی
# =====================================================
df_metrics = df_metrics.sort_values(
    by=["dataset_id", "model"],
    ascending=[True, True]
)

# گرد کردن متریک‌ها
df_metrics = df_metrics.round(3)

# =====================================================
# 9) ذخیره در Excel
# =====================================================
output_path = "/content/drive/MyDrive/results/paper/All_models_metrics_summary.xlsx"
df_metrics.to_excel(output_path, index=False)

print("✅ فایل اکسل ساخته شد:")
print(output_path)

# نمایش
df_metrics
```

# SPEARMAN WITH SCIPY

```python
# @title
# ============================================================
# PEARSON AND SPEARMAN CORRELATION
# FOR FIVE MODELS AND ALL METRICS
# USING SCIPY ONLY
# ============================================================

import pandas as pd
import numpy as np
from scipy.stats import pearsonr, spearmanr
from google.colab import drive

# ============================================================
# 1) اتصال به Google Drive
# ============================================================
drive.mount('/content/drive')

# ============================================================
# 2) خواندن فایل نتایج
# ============================================================
file_path = "/content/drive/MyDrive/results/paper/All_models_metrics_summary.xlsx"
df = pd.read_excel(file_path)

# ============================================================
# 3) تنظیمات
# ============================================================
X_VARIABLE = "ADF"

MODELS = ["TimesFM", "Timer", "Moirai", "Lag_Llama", "Chronos","AutoETS","SNaive","AutoARIMA"]

METRICS = [
    "mae_scaled",
    "rmse_scaled",
    "mse_scaled",
    "mase",
    "crps"
]

ALPHA_05 = 0.05
ALPHA_10 = 0.10

# ============================================================
# 4) بررسی مدل‌های موجود
# ============================================================
available_models = [
    model for model in MODELS
    if model in df["model"].dropna().unique()
]

print("=" * 100)
print("MODELS INCLUDED IN THE ANALYSIS")
print("=" * 100)

for model in available_models:
    print(model)

print(f"\nNumber of models = {len(available_models)}")

# ============================================================
# 5) بررسی Metricهای موجود
# ============================================================
available_metrics = [
    metric for metric in METRICS
    if metric in df.columns
]

print("\n")
print("=" * 100)
print("METRICS INCLUDED IN THE ANALYSIS")
print("=" * 100)

for metric in available_metrics:
    print(metric)

print(f"\nNumber of metrics = {len(available_metrics)}")

# ============================================================
# 6) لیست نتایج
# ============================================================
results = []

# ============================================================
# 7) محاسبه Pearson و Spearman
# ============================================================
for model in MODELS:

    print("\n")
    print("=" * 100)
    print(f"MODEL: {model}")
    print("=" * 100)

    df_model = df[df["model"] == model].copy()

    for metric in available_metrics:

        print("\n" + "-" * 100)
        print(f"Metric: {metric}")
        print("-" * 100)

        # ----------------------------------------------------
        # انتخاب ADF و Metric
        # ----------------------------------------------------
        df_corr = df_model[[X_VARIABLE, metric]].dropna().copy()

        x = df_corr[X_VARIABLE].astype(float).values
        y = df_corr[metric].astype(float).values

        n = len(df_corr)

        print(f"N = {n}")

        # ----------------------------------------------------
        # بررسی تعداد مشاهدات
        # ----------------------------------------------------
        if n < 3:
            print("Not enough observations.")

            results.append({
                "model": model,
                "metric": metric,
                "n": n,

                "pearson_r": np.nan,
                "pearson_p_value": np.nan,
                "pearson_p_0.05": "Not Available",
                "pearson_p_0.10": "Not Available",

                "spearman_rho": np.nan,
                "spearman_p_value": np.nan,
                "spearman_p_0.05": "Not Available",
                "spearman_p_0.10": "Not Available"
            })

            continue

        # ====================================================
        # Pearson با SciPy
        # ====================================================
        try:
            pearson_result = pearsonr(x, y)

            pearson_r = pearson_result.statistic
            pearson_p = pearson_result.pvalue

        except Exception as e:
            print(f"Pearson error: {e}")

            pearson_r = np.nan
            pearson_p = np.nan

        # ====================================================
        # Spearman با SciPy
        # ====================================================
        try:
            spearman_result = spearmanr(x, y)

            spearman_rho = spearman_result.statistic
            spearman_p = spearman_result.pvalue

        except Exception as e:
            print(f"Spearman error: {e}")

            spearman_rho = np.nan
            spearman_p = np.nan

        # ====================================================
        # Pearson significance at α = 0.05
        # ====================================================
        if np.isnan(pearson_p):
            pearson_p_05 = "Not Available"
        elif pearson_p < ALPHA_05:
            pearson_p_05 = "Significant"
        else:
            pearson_p_05 = "Not Significant"

        # ====================================================
        # Pearson significance at α = 0.10
        # ====================================================
        if np.isnan(pearson_p):
            pearson_p_10 = "Not Available"
        elif pearson_p < ALPHA_10:
            pearson_p_10 = "Significant"
        else:
            pearson_p_10 = "Not Significant"

        # ====================================================
        # Spearman significance at α = 0.05
        # ====================================================
        if np.isnan(spearman_p):
            spearman_p_05 = "Not Available"
        elif spearman_p < ALPHA_05:
            spearman_p_05 = "Significant"
        else:
            spearman_p_05 = "Not Significant"

        # ====================================================
        # Spearman significance at α = 0.10
        # ====================================================
        if np.isnan(spearman_p):
            spearman_p_10 = "Not Available"
        elif spearman_p < ALPHA_10:
            spearman_p_10 = "Significant"
        else:
            spearman_p_10 = "Not Significant"

        # ====================================================
        # ذخیره نتایج
        # ====================================================
        results.append({
            "model": model,
            "metric": metric,
            "n": n,

            "pearson_r": pearson_r,
            "pearson_p_value": pearson_p,
            "pearson_p_0.05": pearson_p_05,
            "pearson_p_0.10": pearson_p_10,

            "spearman_rho": spearman_rho,
            "spearman_p_value": spearman_p,
            "spearman_p_0.05": spearman_p_05,
            "spearman_p_0.10": spearman_p_10
        })

        # ====================================================
        # نمایش نتیجه
        # ====================================================
        print(f"Pearson  r  = {pearson_r:.10f}")
        print(f"Pearson  p  = {pearson_p:.10f}")
        print(f"Pearson  p < 0.05 : {pearson_p_05}")
        print(f"Pearson  p < 0.10 : {pearson_p_10}")

        print(f"\nSpearman ρ  = {spearman_rho:.10f}")
        print(f"Spearman p  = {spearman_p:.10f}")
        print(f"Spearman p < 0.05 : {spearman_p_05}")
        print(f"Spearman p < 0.10 : {spearman_p_10}")

# ============================================================
# 8) ساخت DataFrame نهایی
# ============================================================
results_df = pd.DataFrame(results)

# ============================================================
# 9) گرد کردن مقادیر عددی
# ============================================================
results_df[
    [
        "pearson_r",
        "pearson_p_value",
        "spearman_rho",
        "spearman_p_value"
    ]
] = results_df[
    [
        "pearson_r",
        "pearson_p_value",
        "spearman_rho",
        "spearman_p_value"
    ]
].round(10)

# ============================================================
# 10) نمایش جدول نهایی
# ============================================================
print("\n")
print("=" * 120)
print("FINAL CORRELATION RESULTS")
print("=" * 120)

print(results_df.to_string(index=False))

# ============================================================
# 11) ذخیره تمام نتایج در Excel
# ============================================================
output_path = (
    "/content/drive/MyDrive/results/paper"
    "Pearson_Spearman_Five_Models_With_PValues.xlsx"
)

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:

    # --------------------------------------------------------
    # تمام نتایج
    # --------------------------------------------------------
    results_df.to_excel(
        writer,
        sheet_name="All_Results",
        index=False
    )

    # --------------------------------------------------------
    # نتایج معنادار در α = 0.05
    # --------------------------------------------------------
    significant_05 = results_df[
        (results_df["pearson_p_value"] < ALPHA_05)
        |
        (results_df["spearman_p_value"] < ALPHA_05)
    ].copy()

    significant_05.to_excel(
        writer,
        sheet_name="Significant_0.05",
        index=False
    )

    # --------------------------------------------------------
    # نتایج معنادار در α = 0.10
    # --------------------------------------------------------
    significant_10 = results_df[
        (results_df["pearson_p_value"] < ALPHA_10)
        |
        (results_df["spearman_p_value"] < ALPHA_10)
    ].copy()

    significant_10.to_excel(
        writer,
        sheet_name="Significant_0.10",
        index=False
    )

print("\n")
print("=" * 100)
print("EXCEL FILE SAVED")
print("=" * 100)

print(output_path)
```
